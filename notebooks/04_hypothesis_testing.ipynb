# Hypothesis Testing
Statistical hypothesis tests on both datasets including t-tests, ANOVA, and Chi-square tests.

```python
# Imports & Setup

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
print(":white_check_mark: Hypothesis Testing Framework Ready")
```

```python
# Independent T-Test Function

def independent_ttest(group1, group2, label1="Group 1", label2="Group 2"):
    """
    Perform independent samples t-test
    Parameters:
    - group1, group2: Data arrays/series
    - label1, label2: Group names
    Returns:
    - Dictionary with test results
    """
    t_stat, pvalue = stats.ttest_ind(group1, group2)
    return {
        'group_1': label1,
        'group_2': label2,
        't_statistic': round(t_stat, 4),
        'p_value': round(pvalue, 6),
        'significant': 'Yes' if pvalue < 0.05 else 'No',
        'mean_1': round(group1.mean(), 2),
        'mean_2': round(group2.mean(), 2)
    }
print(":white_check_mark: T-test function defined")
```

```python
# One-Way ANOVA Function

def anova_test(*groups, group_labels=None):
    """
    Perform one-way ANOVA across multiple groups
    Parameters:
    - *groups: Multiple data arrays/series
    - group_labels: Names of groups
    Returns:
    - Dictionary with ANOVA results
    """
    f_stat, pvalue = stats.f_oneway(*groups)
    return {
        'test': 'One-Way ANOVA',
        'f_statistic': round(f_stat, 4),
        'p_value': round(pvalue, 6),
        'significant': 'Yes' if pvalue < 0.05 else 'No',
        'num_groups': len(groups)
    }
print(":white_check_mark: ANOVA function defined")
```

```python
# Chi-Square Test

def chi_square_test(contingency_table, label=""):
    """
    Perform chi-square test for independence
    Parameters:
    - contingency_table: 2D array of observed frequencies
    - label: Test label
    Returns:
    - Dictionary with chi-square results
    """
    chi2, pvalue, dof, expected = stats.chi2_contingency(contingency_table)
    return {
        'test': label if label else 'Chi-Square Test',
        'chi2_statistic': round(chi2, 4),
        'p_value': round(pvalue, 6),
        'df': dof,
        'significant': 'Yes' if pvalue < 0.05 else 'No'
    }
print(":white_check_mark: Chi-square test function defined")
```

```python
# Normality Test

def normality_test(data, label=""):
    """
    Perform Shapiro-Wilk test for normality
    Parameters:
    - data: Data array/series
    - label: Data label
    Returns:
    - Dictionary with normality test results
    """
    stat, pvalue = stats.shapiro(data)
    return {
        'variable': label if label else 'Variable',
        'statistic': round(stat, 4),
        'p_value': round(pvalue, 6),
        'normal': 'Yes' if pvalue > 0.05 else 'No'
    }
print(":white_check_mark: Normality test function defined")
```

```python
# Z-Test Function

def ztest_groups(group1, group2, label1="Group 1", label2="Group 2"):
    """
    Perform z-test to compare two groups (for large samples)
    Parameters:
    - group1, group2: Data arrays/series
    - label1, label2: Group names
    Returns:
    - Dictionary with z-test results
    """
    from scipy.stats import norm
    # Calculate means and standard deviations
    mean1 = group1.mean()
    mean2 = group2.mean()
    n1 = len(group1)
    n2 = len(group2)
    # Standard error of the difference
    se1 = group1.std() / np.sqrt(n1)
    se2 = group2.std() / np.sqrt(n2)
    se_diff = np.sqrt(se1**2 + se2**2)
    # Calculate z-statistic
    z_stat = (mean1 - mean2) / se_diff if se_diff != 0 else 0
    # Two-tailed p-value
    pvalue = 2 * (1 - norm.cdf(abs(z_stat)))
    return {
        'group_1': label1,
        'group_2': label2,
        'z_statistic': round(z_stat, 4),
        'p_value': round(pvalue, 6),
        'significant': 'Yes' if pvalue < 0.05 else 'No',
        'mean_1': round(mean1, 2),
        'mean_2': round(mean2, 2),
        'difference': round(mean1 - mean2, 2)
    }
print(" Z-test function defined")

```

## Student Data Hypothesis Testing

```python
#Load Student Data

df_student = pd.read_csv('../data/ai_assistant_usage_student_life.csv')
print("\n" + "="*60)
print("STUDENT DATA - HYPOTHESIS TESTS")
print("="*60)
print(f"Dataset shape: {df_student.shape}")
print(f"Numeric columns: {df_student.select_dtypes(include=[np.number]).columns.tolist()}")
print(f"Categorical columns: {df_student.select_dtypes(include='object').columns.tolist()}")
```

```python
# Normality Tests - Student Data

print("\n" + "-"*60)
print("Normality Tests (Shapiro-Wilk)")
print("-"*60)
numeric_cols = df_student.select_dtypes(include=[np.number]).columns
normality_results = []
for col in numeric_cols:
    result = normality_test(df_student[col].dropna(), label=col)
    normality_results.append(result)
normality_df = pd.DataFrame(normality_results)
print(normality_df.to_string(index=False))
```

## Workforce Data Hypothesis Testing

```python
# Load Workforce Data

df_workforce = pd.read_csv('../data/global_ai_workforce_automation_2015_2025.csv')
print("\n" + "="*60)
print("WORKFORCE DATA - HYPOTHESIS TESTS")
print("="*60)
print(f"Dataset shape: {df_workforce.shape}")
print(f"Numeric columns: {df_workforce.select_dtypes(include=[np.number]).columns.tolist()}")
print(f"Categorical columns: {df_workforce.select_dtypes(include='object').columns.tolist()}")
```

```python
# Normality Tests - Workforce Data

print("\n" + "-"*60)
print("Normality Tests (Shapiro-Wilk)")
print("-"*60)
numeric_cols = df_workforce.select_dtypes(include=[np.number]).columns
normality_results = []
for col in numeric_cols:
    result = normality_test(df_workforce[col].dropna(), label=col)
    normality_results.append(result)
normality_df = pd.DataFrame(normality_results)
print(normality_df.to_string(index=False))
```

```python
# Summary

print("\n" + "="*60)
print(":white_check_mark: HYPOTHESIS TESTING FRAMEWORK COMPLETE")
print("="*60)
print("\nAll test functions ready:")
print("  :white_check_mark: Independent t-tests")
print("  :white_check_mark: One-way ANOVA")
print("  :white_check_mark: Chi-square tests")
print("  :white_check_mark: Normality tests")
print("\nReady for detailed hypothesis testing and visualization!")
```

Test: One-Way ANOVA

```python
print("\n" + "="*60)
print("H1: Does AI Assistance Level affect Satisfaction?")
print("="*60)
print("\nH₀: All AI levels have SAME satisfaction")
print("H₁: AI levels have DIFFERENT satisfaction")
level_1 = df_student[df_student['AI_AssistanceLevel']==1]['SatisfactionRating']
level_2 = df_student[df_student['AI_AssistanceLevel']==2]['SatisfactionRating']
level_3 = df_student[df_student['AI_AssistanceLevel']==3]['SatisfactionRating']
level_4 = df_student[df_student['AI_AssistanceLevel']==4]['SatisfactionRating']
level_5 = df_student[df_student['AI_AssistanceLevel']==5]['SatisfactionRating']
result_h1 = anova_test(level_1, level_2, level_3, level_4, level_5)
print(f"\nTest: One-Way ANOVA")
print(f"F-statistic: {result_h1['f_statistic']}")
print(f"P-value: {result_h1['p_value']}")
print(f"Significant: {result_h1['significant']}")
if result_h1['p_value'] < 0.05:
    print(f"\n:white_check_mark: REJECT H₀: AI assistance DOES affect satisfaction!")
else:
    print(f"\n:x: FAIL TO REJECT H₀")
print(f"\nMeans: L1={level_1.mean():.2f}, L2={level_2.mean():.2f}, L3={level_3.mean():.2f}, L4={level_4.mean():.2f}, L5={level_5.mean():.2f}")
```

Test: Independent t-test

```python
print("\n" + "="*60)
print("H2: Do Task Completers have higher satisfaction?")
print("="*60)
print("\nH₀: Completers = Non-completers in satisfaction")
print("H₁: Completers ≠ Non-completers in satisfaction")
completed = df_student[df_student['FinalOutcome']=='Assignment Completed']['SatisfactionRating']
not_completed = df_student[df_student['FinalOutcome']!='Assignment Completed']['SatisfactionRating']
result_h2 = independent_ttest(completed, not_completed, 'Completed', 'Not Completed')
print(f"\nTest: Independent t-test")
print(f"t-statistic: {result_h2['t_statistic']}")
print(f"P-value: {result_h2['p_value']}")
print(f"Significant: {result_h2['significant']}")
if result_h2['p_value'] < 0.05:
    print(f"\n:white_check_mark: REJECT H₀: Completion DOES affect satisfaction!")
else:
    print(f"\n:x: FAIL TO REJECT H₀")
print(f"\nMeans: Completed={result_h2['mean_1']}, Not Completed={result_h2['mean_2']}, Difference={result_h2['mean_1']-result_h2['mean_2']:.2f}")
```

Test: Z-test

```python
print("\n" + "="*60)
print("H3: Does AI Investment increase Job Creation?")
print("="*60)
print("\nH₀: AI investment has NO effect on job creation")
print("H₁: Higher AI investment leads to MORE job creation")
median_inv = df_workforce['AI_Investment_BillionUSD'].median()
high_inv = df_workforce[df_workforce['AI_Investment_BillionUSD'] >= median_inv]['Job_Creation_Million']
low_inv = df_workforce[df_workforce['AI_Investment_BillionUSD'] < median_inv]['Job_Creation_Million']
result_h3 = ztest_groups(high_inv, low_inv, f'High Investment (≥${median_inv:.1f}B)', f'Low Investment (<${median_inv:.1f}B)')
print(f"\nTest: Z-test")
print(f"Z-statistic: {result_h3['z_statistic']}")
print(f"P-value: {result_h3['p_value']}")
print(f"Significant: {result_h3['significant']}")
if result_h3['p_value'] < 0.05:
    print(f"\n:white_check_mark: REJECT H₀: AI investment DOES increase job creation!")
else:
    print(f"\n:x: FAIL TO REJECT H₀")
print(f"\nMeans: High={result_h3['mean_1']}, Low={result_h3['mean_2']}, Difference={result_h3['difference']}")
```

Test: Z-test

```python
print("\n" + "="*60)
print("H4: Does higher automation reduce employment rates?")
print("="*60)
print("\nH₀: Automation has NO effect on employment")
print("H₁: Higher automation REDUCES employment")
median_auto = df_workforce['Automation_Rate_Percent'].median()
high_auto = df_workforce[df_workforce['Automation_Rate_Percent'] >= median_auto]['Employment_Rate_Percent']
low_auto = df_workforce[df_workforce['Automation_Rate_Percent'] < median_auto]['Employment_Rate_Percent']
result_h4 = ztest_groups(high_auto, low_auto, f'High Automation (≥{median_auto:.1f}%)', f'Low Automation (<{median_auto:.1f}%)')
print(f"\nTest: Z-test")
print(f"Z-statistic: {result_h4['z_statistic']}")
print(f"P-value: {result_h4['p_value']}")
print(f"Significant: {result_h4['significant']}")
if result_h4['p_value'] < 0.05:
    print(f"\n:white_check_mark: REJECT H₀: Automation DOES reduce employment!")
else:
    print(f"\n:x: FAIL TO REJECT H₀")
print(f"\nMeans: High Automation={result_h4['mean_1']}%, Low Automation={result_h4['mean_2']}%, Difference={result_h4['difference']:.2f}%")
```

```python
print("\n" + "="*70)
print("HYPOTHESIS TESTING SUMMARY")
print("="*70)
print(f"\nH1 (ANOVA): {result_h1['significant']}")
print(f"H2 (t-test): {result_h2['significant']}")
print(f"H3 (Z-test): {result_h3['significant']}")
print(f"H4 (Z-test): {result_h4['significant']}")
print("\n:white_check_mark: All hypothesis tests complete!")
```
