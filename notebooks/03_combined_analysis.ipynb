
# Combined Data Summary

Overview of both datasets and integration points.


```python
import pandas as pd
import numpy as np

# Load both datasets
df_student = pd.read_csv('../data/ai_assistant_usage_student_life.csv')
df_workforce = pd.read_csv('../data/global_ai_workforce_automation_2015_2025.csv')

print("DATASET COMPARISON")

print(f"\n Student AI Usage Data:")
print(f"  - Shape: {df_student.shape}")
print(f"  - Rows: {len(df_student)}, Columns: {len(df_student.columns)}")

print(f"\n Global AI Workforce Automation Data:")
print(f"  - Shape: {df_workforce.shape}")
print(f"  - Rows: {len(df_workforce)}, Columns: {len(df_workforce.columns)}")
```

```python
# Data Quality

print("DATA QUALITY SUMMARY")

print("\n Student Data - Missing Values:")
missing_student = df_student.isnull().sum()
if missing_student.sum() > 0:
    print(missing_student[missing_student > 0])
else:
    print("  No missing values ✅")

print("\n Workforce Data - Missing Values:")
missing_workforce = df_workforce.isnull().sum()
if missing_workforce.sum() > 0:
    print(missing_workforce[missing_workforce > 0])
else:
    print("  No missing values ✅")
```

```python
# Numeric Summary

print("NUMERIC COLUMNS SUMMARY")

print("\n Student Data - Numeric Columns:")
student_numeric = df_student.select_dtypes(include=[np.number]).columns.tolist()
print(f"  {student_numeric}")

print("\n Workforce Data - Numeric Columns:")
workforce_numeric = df_workforce.select_dtypes(include=[np.number]).columns.tolist()
print(f"  {workforce_numeric}")
```
